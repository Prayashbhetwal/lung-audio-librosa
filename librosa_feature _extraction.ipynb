```python
#calaculate the amplitude envelope
frame_size = 1024
hop_length =512
#Calculating the amplitude envelope
def amplitude_envelope(prayash, frame_size, hop_length):
  return np.array([max(prayash[i:i+frame_size]) for i in range(0, prayash.size, hop_length)])
```

```python
ae=amplitude_envelope(prayash, frame_size, hop_length)
print(ae)
ae.ndim
```

```python
hop_length=512
frames=range(len(ae))
t=librosa.frames_to_time(frames,hop_length=hop_length)
plt.figure(figsize=(15,17))
plt.subplot(3,1,1)
librosa.display.waveshow(prayash, alpha=0.5)
plt.plot(t ,ae,color="r")
plt.title("AMPLITUDE ENVELOPE")
plt.ylim(-1,1)
plt.show()
```

```python
#RMSE
def rms(signal, frame_length,hop_length):
    rms=[]
    for i in range(0,len(prayash),hop_length):
        rms_current_frame=np.sqrt(np.sum(prayash[i:i+frame_length]**2)/frame_length)
        rms.append(rms_current_frame)
    return np.array(rms)
```

```python
Frame_length = 1024
Hop_length = 512

# Assuming 'prayash' is  audio signal
rmsprayash = rms(prayash, frame_length=Frame_length, hop_length=Hop_length)
print(rmsprayash)
```

```python
plt.figure(figsize=(15,17))
plt.subplot(3,1,1)
librosa.display.waveshow(prayash,sr=sr, alpha=0.5)
plt.plot(t ,rmsprayash,color="r")
plt.plot(t ,rmsprayash,color="y")
plt.title("RMS")
plt.ylim(-1,1)
plt.show()
```

```python
#fourier transform to now change frequency 
prayash,sr=librosa.load(prayash_file)
ft = sp.fft.fft(prayash)
magnitude = np.absolute(ft)
frequency = np.linspace(0, sr, len(magnitude)) 

```

```python
plt.figure(figsize=(18, 8))
plt.plot(frequency[:5000], magnitude[:5000]) # magnitude spectrum
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()
```

```python
samples = range(len(prayash))
t = librosa.samples_to_time(samples, sr=sr)

plt.figure(figsize=(18, 8))
plt.plot(t[10000:10400], prayash[10000:10400])
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
#features for beat tempo analysis 
beattempo=librosa.beat.tempo(y=prayash, sr=sr)
print(beattempo)
#to calculate onset envelope
o_env = librosa.onset.onset_strength(y=prayash, sr=sr)
o_env.size
#for tempogram
tempogram=librosa.feature.tempogram(y=prayash,sr=sr,onset_envelope=o_env,)
print(tempogram)
#forier tempogram
hop_length=512
forier_tempogram= librosa.feature.fourier_tempogram(onset_envelope=o_env, sr=sr, hop_length=hop_length)
print(forier_tempogram)
#tempogram ratio
tgr = librosa.feature.tempogram_ratio(tg=tempogram, sr=sr)
print(tgr.flatten())





```

```python
#spectral featrues 
mel_spectrogram=librosa.feature.melspectrogram(y=prayash, sr=sr)
print(mel_spectrogram)
#short time forier transformation
stft=librosa.feature.chroma_stft(y=prayash, sr=sr)
print(stft)
#sconstant q transform
chroma_cq = librosa.feature.chroma_cqt(y=prayash, sr=sr)
print(chroma_cq)
#chroma feature with energy
chroma_energy=chroma_cens = librosa.feature.chroma_cens(y=prayash, sr=sr)
print(chroma_energy)
#mel_frequency cepstral coefficent
mfcc=librosa.feature.mfcc(y=prayash, sr=sr)
print(mfcc)









```

```python
prayash,sr=librosa.load(prayash_file)
pitches, magnitudes = librosa.piptrack(y=prayash, sr=sr)
pitch_max = []
for t in range(pitches.shape[1]):  # Iterate over time frames
    index = magnitudes[:, t].argmax()  # Find index with max magnitude
    pitch_max.append(pitches[index, t])  # Get the corresponding pitch

# Convert to a NumPy array
pitch_max = np.array(pitch_max)

# Plotting the pitch contour
plt.figure(figsize=(10, 6))
plt.plot(librosa.times_like(pitch_max), pitch_max, label='Estimated Pitch', color='r')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.title('Pitch Estimation using librosa.piptrack')
plt.legend()
plt.show()
```
